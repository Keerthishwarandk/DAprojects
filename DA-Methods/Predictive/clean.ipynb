```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Load your dataset
df = pd.read_excel("datasetcsv(new).xlsx", engine="openpyxl")

# Generate random dates between Jan 1, 2024 and Dec 31, 2025
start_date = datetime(2024, 1, 1)
end_date = datetime(2025, 12, 31)
date_range = (end_date - start_date).days

# Assign random dates
df['Date'] = [start_date + timedelta(days=np.random.randint(0, date_range)) for _ in range(len(df))]

# Optional: Format the date column
df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')


```

```python
#changing to 2 decimal values
#A lambda in Python is an anonymous function — meaning it doesn’t have a name.
# It’s used when you need a small function for a short time.

df['Price_per_Litre'] = df['Price_per_Litre'].map(lambda x: f"{x:.2f}")
df['Total_Amount'] = df['Total_Amount'].map(lambda x: f"{x:.2f}")
df['Quantity_Liters'] = df['Quantity_Liters'].map(lambda x: f"{x:.2f}")
```

```python

#changing to lowercase and removing spaces
df['LocationID'] = df['LocationID'].str.strip().str.lower()
df['City'] = df['City'].str.strip().str.lower()
df['FuelType'] = df['FuelType'].str.strip().str.lower()
df['VehicleType'] = df['VehicleType'].str.strip().str.lower()
```

```python
# Mapping 1 to 'Yes' and 0 to 'No'
df['LoyaltyCardUsed'] = df['LoyaltyCardUsed'].map({1: 'Yes', 0: 'No'})
```

```python
# Save the updated dataset
df.to_csv("final_dataset.csv", index=False)
```
